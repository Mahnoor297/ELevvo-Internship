### Important info of Unique Users

```python
Sales_grouped_data = df.groupby('User_ID').agg(
                        total_order = ("User_ID","count"),
                        total_amount_spent = ("Purchase","sum"),
                        average_amount_spent = ("Purchase","mean")
                        ).reset_index()

Sales_grouped_data['User_ID'] = Sales_grouped_data['User_ID'].astype(str)
Sales_grouped_data
```

### Our Top 10 Users

```python
plt.figure(figsize=(12,5))

ax=sns.barplot(data=Sales_grouped_data.sort_values(by="total_order", ascending=False).head(10), x='User_ID', y='total_order')

plt.title("Top 10 frequent users")
plt.bar_label(*ax.containers)
plt.xlabel("User_ID")
plt.ylabel("Total Orders")
plt.show()
```

##### User with ID 1001680 has most number of orders i.e. 1026

## Grouped Data w.r.t Unique Users

```python
grouped_data=df.groupby("User_ID").agg({
    "Product_ID":lambda x:",".join(map(str,set(x))),
    "Gender":"first",
    "Age":"first",
    "Occupation":"first",
    "City_Category":"first",
    "Stay_In_Current_City_Years":"first",
    "Marital_Status":"first",
    "Product_Category_1":lambda x:",".join(map(str,set(x.dropna()))),
    "Product_Category_2": lambda x:",".join(map(str,set(x.dropna()))),
    "Product_Category_3":lambda x:",".join(map(str,set(x.dropna()))),
    "Purchase":sum
}).reset_index()
grouped_data.head()
```

```python
grouped_data.info()
```

## EDA

```python
user_product_counts=df.groupby("User_ID")["Product_ID"].nunique()

max_products=user_product_counts.max()
min_products=user_product_counts.min()

user_with_max_products= user_product_counts[user_product_counts==max_products].index.tolist()
user_with_min_products= user_product_counts[user_product_counts==min_products].index.tolist()

print(f"Maximum Products bought by a single User: {max_products}")
print(f"User who bought maximum products: {user_with_max_products}")
print(f"Min Products bought by a single User: {min_products}")
print(f"User who bought minimum products: {user_with_min_products}")
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(user_product_counts, bins=20, kde=True, color="skyblue")  # KDE for smooth distribution curve
plt.xlabel("Number of Products Bought")
plt.ylabel("Number of Users")
plt.title("Distribution of Products Bought per User")
plt.grid(axis='y', linestyle="--", alpha=0.7)
plt.show()
```

### Right-Skewed Distribution:

##### 1) Most users have purchased only a few products.
##### 2) A small number of users have purchased a large number of products.


### Majority of Users Buy Few Products:

##### 1) The highest bar is on the left, meaning most users bought very few products (probably between 1-50 products).
##### 2) The number of users decreases rapidly as the number of products increases.


### Long Tail Effect:

##### 1) There are some users who bought a significantly higher number of products (over 600–1000 products).
##### 2) These are outliers or high-value customers.


### Possible Business Insights:
##### 1) Most users are occasional buyers, meaning they purchase only a few items.
##### 2) A small percentage of users are heavy buyers, which might indicate loyal customers or bulk buyers.


##### If the goal is to increase customer purchases, marketing efforts should focus on converting low-frequency buyers into repeat customers.

```python
plt.figure(figsize=(8,5))

# Plot histogram with KDE
sns.histplot(df["Purchase"], bins=50, kde=True, kde_kws={'bw_adjust': 1.5}, color="royalblue")

# Add mean and median lines
mean_value = df["Purchase"].mean()
median_value = df["Purchase"].median()
plt.axvline(mean_value, color='red', linestyle='dashed', label=f'Mean: {mean_value:.2f}')
plt.axvline(median_value, color='green', linestyle='dashed', label=f'Median: {median_value:.2f}')
plt.legend()

# Titles and labels
plt.title("Distribution of Purchase Amounts", fontsize=14)
plt.xlabel("Purchase Amount", fontsize=12)
plt.ylabel("Frequency", fontsize=12)
plt.grid(axis="y", linestyle="--", alpha=0.7)

plt.show()
```

##### The majority of purchases are moderate, but a small number of high-value purchases significantly influence overall sales.
